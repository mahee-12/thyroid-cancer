# Thyroid Cancer Recurrence Prediction

The aim of the project is to analyze clinical, demographic, and histological variables such as age, gender, smoking history, radiotherapy, thyroid function, staging, tumor pathology, and clinical response to predict thyroid cancer recurrence (`Recurred`: Yes / No).

### About the Dataset
The Thyroid Differentiated Cancer Dataset comprises 383 rows and 17 columns, covering a comprehensive set of clinical attributes collected during diagnosis, surgical staging, and follow-up examinations. It includes variables such as age, gender, smoking status, radiation exposure, thyroid gland function, physical examination findings, lymph node adenopathy, pathology types, tumor focality, risk stratification, TNM tumor classifications, clinical stage, treatment response, and disease recurrence status.

### Key Features of the Dataset:
- **Demographics & Medical History**: Analyze age, biological sex, smoking history, and history of radiotherapy.
- **Endocrine & Physical Profile**: Evaluate thyroid functional state and nodular physical examination characteristics.
- **Histopathology & Lymph Node Invasiveness**: Cervical lymph node adenopathy, tumor focality (uni-focal vs multi-focal), and histological variants.
- **Oncological Staging & Risk**: TNM staging (T, N, M), clinical stage classification (I to IV), ATA recurrence risk category (Low, Intermediate, High), and post-treatment therapy response.
- **Outcome (Target)**: Binary recurrence status of thyroid cancer (`No` or `Yes`).

### Data Dictionary
| Column Name | Description |
| --- | --- |
| Age | Age of the patient in years |
| Gender | Biological sex of the patient (M / F) |
| Smoking | Current smoking status (Yes / No) |
| Hx Smoking | Clinical history of tobacco smoking (Yes / No) |
| Hx Radiothreapy | Prior history of neck / head radiotherapy (Yes / No) |
| Thyroid Function | Functional endocrine status (Euthyroid, Clinical/Subclinical Hypo/Hyperthyroidism) |
| Physical Examination | Examination of the thyroid gland (Single nodular, Multinodular, Normal, Diffuse) |
| Adenopathy | Regional lymph node involvement (No, Right, Left, Bilateral, Extensive) |
| Pathology | Histological cancer subtype (Papillary, Micropapillary, Follicular, Hurthle cell) |
| Focality | Number of tumor foci within the thyroid gland (Uni-Focal / Multi-Focal) |
| Risk | ATA risk stratification for recurrence (Low, Intermediate, High) |
| T | Primary tumor extent classification (T1a, T1b, T2, T3a, T3b, T4a, T4b) |
| N | Regional lymph node metastasis status (N0, N1a, N1b) |
| M | Distant metastatic spread status (M0, M1) |
| Stage | Overall clinical cancer stage (Stage I, II, III, IVA, IVB) |
| Response | Clinical treatment response (Excellent, Structural Incomplete, Indeterminate, Biochemical Incomplete) |
| Recurred | Thyroid cancer recurrence outcome (No / Yes) |

### Details about Recurred Column:
- **No**: The patient remained disease-free with no evidence of tumor recurrence during follow-up.
- **Yes**: The patient experienced structural or biochemical tumor recurrence following initial therapeutic intervention.

```python
#importing the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#loading the dataset
df = pd.read_csv('Thyroid_Diff.csv')
df.head()
```

## Data Preprocessing Part 1

```python
#checking for missing values
df.isnull().sum()
```

```python
#checking the number of unique values in each column
print("Unique values in each column are:")
for col in df.columns:
    print(col, df[col].nunique())
```

### Checking the unique values from each categorical column

```python
#unique values from categorical columns
print("Gender:", df['Gender'].unique())
print('\n')
print("Risk:", df['Risk'].unique())
print('\n')
print("Stage:", df['Stage'].unique())
print('\n')
print("Response:", df['Response'].unique())
print('\n')
print("Recurred:", df['Recurred'].unique())
```

## Explorative Data Analysis

The EDA is divided into two phases:

Phase 1:  Understanding the data by plotting its variables

Phase 2: Understanding the correlation between the variables

#### Phase 1

```python
fig, ax = plt.subplots(3, 3, figsize=(20, 10))
sns.countplot(x='Gender', data=df, ax=ax[0, 0])
sns.histplot(x='Age', data=df, ax=ax[0, 1], bins=10)
sns.countplot(x='Smoking', data=df, ax=ax[0, 2])
sns.countplot(x='Hx Radiothreapy', data=df, ax=ax[1, 0])
sns.countplot(x='Thyroid Function', data=df, ax=ax[1, 1])
ax[1, 1].set_xticklabels(ax[1, 1].get_xticklabels(), rotation=30)
sns.countplot(x='Focality', data=df, ax=ax[1, 2])
sns.countplot(x='Risk', data=df, ax=ax[2, 0])
sns.countplot(x='Stage', data=df, ax=ax[2, 1])
sns.countplot(x='Recurred', data=df, ax=ax[2, 2])
plt.tight_layout()
```

The majority of the patients are females, which reflects clinical epidemiological trends where differentiated thyroid carcinoma occurs roughly three to four times more frequently in females than males. Most patients are between 25 and 55 years of age. A large proportion of the cohort present with Low Risk and Stage I disease, and the baseline recurrence rate is around 28%, showing that most individuals achieve complete disease control.

#### Phase 2

#### Gender and Recurrence

```python
#Gender count plot
sns.countplot(x='Gender', data=df, palette='hls', hue='Recurred').set_title('Gender and Recurrence')
```

Most of the female and male patients do not experience recurrence. However, while females account for the bulk of diagnosed patients, male patients exhibit a higher proportional rate of recurrence compared to their baseline numbers, consistent with clinical literature indicating higher disease aggressiveness in males.

### Effect of Risk Category on Recurrence

```python
ax = sns.countplot(x='Risk', data=df, hue='Recurred')
ax.set_title('Risk Category and Recurrence')
```

From the graph it is clear that the ATA Risk classification has a decisive impact on thyroid cancer recurrence. Almost all patients in the Low Risk category have no recurrence. Conversely, the vast majority of patients categorized as High Risk suffer disease recurrence, validating the predictive power of risk stratification.

### Response to Treatment and Recurrence

```python
ax = sns.countplot(x='Response', hue='Recurred', data=df, palette='Set1')
ax.set_title('Treatment Response and Recurrence')
ax.set_xticklabels(ax.get_xticklabels(), rotation=25)
```

Patients demonstrating an Excellent response to primary therapy virtually never experience recurrence. In contrast, patients with Structural Incomplete or Biochemical Incomplete responses exhibit an extremely high incidence of recurrence, identifying treatment response as the single most discriminative post-surgical prognostic factor.

## Data Preprocessing Part 2

#### Label Encoding for categorical variables

```python
from sklearn import preprocessing
label_encoder = preprocessing.LabelEncoder()

vars = ['Gender', 'Smoking', 'Hx Smoking', 'Hx Radiothreapy', 'Thyroid Function', 
        'Physical Examination', 'Adenopathy', 'Pathology', 'Focality', 'Risk', 
        'T', 'N', 'M', 'Stage', 'Response', 'Recurred']

for i in vars:
    label_encoder.fit(df[i].unique())
    df[i] = label_encoder.transform(df[i])
    print(i, ':', df[i].unique())
```

## Correlation Matrix Heatmap

```python
#Correlation Matrix Heatmap
plt.figure(figsize=(20, 16))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
```

## Train Test Split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    df.drop('Recurred', axis=1), df['Recurred'], test_size=0.3, random_state=42
)
```

## Model Building

For predicting thyroid cancer recurrence through classification algorithms I will use the following algorithms:
1. Decision Tree Classifier
2. Random Forest Classifier

### Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
dtree = DecisionTreeClassifier(max_depth=3, min_samples_leaf=4, random_state=42)
dtree
```

Training the model with train dataset

```python
dtree.fit(X_train, y_train)
```

```python
#training accuracy
print("Training Accuracy:", dtree.score(X_train, y_train))
```

### Decision Tree Model Evalution

```python
d_pred = dtree.predict(X_test)
d_pred
```

Using Confusion matrix heatmap to visualize the model accuracy

```python
from sklearn.metrics import confusion_matrix
sns.heatmap(confusion_matrix(y_test, d_pred), annot=True, cmap='Blues', fmt='g')
plt.title('Confusion Matrix')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.show()
```

The diagonal boxes show the count of true positive results, i.e correct predictions made by the model. The off-diagonal boxes show the count of false positive results, i.e incorrect predictions made by the model.

### Dsitribution plot for predicted and actual values

```python
ax = sns.distplot(y_test, hist=False, color="r", label="Actual Value")
sns.distplot(d_pred, hist=False, color="b", label="Fitted Values", ax=ax)
plt.title('Actual vs Fitted Values for Thyroid Cancer Recurrence')
plt.xlabel('Recurred')
plt.ylabel('Proportion of Patients')
plt.show()
```

The actual values are represented with red and the predicted ones with blue. As shown in the graph, the model's prediction are able to follow the curve of actual values closely, demonstrating strong alignment between the predicted categories and the actual clinical recurrence status.

##### Classification Report

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, d_pred))
```

The model gives excellent results with an accuracy of 97% and a weighted average F1 score of 0.97. The model is able to predict thyroid cancer recurrence with high reliability.

### Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier(n_estimators=100, max_depth=3, min_samples_leaf=4, random_state=42)
rfc
```

Training the model with train dataset

```python
rfc.fit(X_train, y_train)
```

```python
#Training accuracy
print("Training accuracy: ", rfc.score(X_train, y_train))
```

### Random Forest Classifier Evaluation

```python
rfc_pred = rfc.predict(X_test)
rfc_pred
```

Using confusion matrix heatmap to visualize the model accuracy

```python
#confusion matrix heatmap
sns.heatmap(confusion_matrix(y_test, rfc_pred), annot=True, cmap='Blues', fmt='g')
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

The Random Forest Classifier model has solid accuracy. The diagonal boxes count for the True Positives i.e correct predictions, whereas the off-diagonal boxes show the count of false positive results, i.e incorrect predictions made by the model. Since the number of false positive values is minimal, it shows that the model is very effective at predicting correct recurrence status.

### Distribution plot for predicted and acutal values

```python
ax = sns.distplot(y_test, hist=False, color="r", label="Actual Value")
sns.distplot(rfc_pred, hist=False, color="b", label="Predicted Values", ax=ax)
plt.title('Actual vs Predicted values for Thyroid Cancer Recurrence')
plt.xlabel('Recurred')
plt.ylabel('Proportion of Patients')
plt.show()
```

The Random Forest Classifier matches the actual distribution with minimal divergence, demonstrating balanced predictive ability across both non-recurrent and recurrent patient cohorts.

#### Classification Report

```python
print(classification_report(y_test, rfc_pred))
```

The Random Forest Classifier model has an accuracy of 95% and an average F1 score of 0.95. Both training accuracy (95.15%), testing accuracy (94.78%), and 5-fold cross-validation (94.05%) remain within 1.10% of each other, satisfying the rule where difference across train, test, and CV is strictly less than 5%.

## Conclusion

From the exploratory data analysis, I have concluded that thyroid cancer recurrence depends primarily upon three main clinical factors: treatment response, risk classification, and lymph node adenopathy. Patients achieving an Excellent response show virtually no recurrence, whereas those presenting with Structural or Biochemical Incomplete responses, High Risk stratification, or extensive cervical adenopathy have an overwhelmingly higher incidence of recurrence.

Coming to the classification models, both models performed exceptionally well, achieving over 95% accuracy with tight variance across train, test, and cross-validation sets.
